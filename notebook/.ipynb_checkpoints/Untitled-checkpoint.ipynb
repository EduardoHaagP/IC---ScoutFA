```python
import nflreadpy as nfl
import polars as pl

seasons = [2023, 2024]

pbp = nfl.load_pbp(seasons)
part = nfl.load_participation(seasons)

part_cols = part.select([
    "nflverse_game_id",
    "play_id",
    "offense_formation",
    "offense_personnel",
    "defense_personnel",
    "defenders_in_box",
    "number_of_pass_rushers",
])

pbp_full = pbp.join(
    part_cols,
    on=["nflverse_game_id", "play_id"],
    how="left",
)

# Exemplo: jogadas de passe com 11 personnel em shotgun
df = pbp_full.filter(
    (pl.col("play_type") == "pass")
    & (pl.col("offense_personnel").str.contains("1 RB, 1 TE, 3 WR"))
    & (pl.col("offense_formation") == "SHOTGUN")
)
print(df)
```
